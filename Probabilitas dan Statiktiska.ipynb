```python
import re
import pandas as pd
import numpy as np
from scipy import stats
from google.colab import files

print("Library berhasil diimpor!")
```

```python
uploaded = files.upload()

nama_file = list(uploaded.keys())[0]

print("File terbaca:", nama_file)
```

```python
df = pd.read_excel(nama_file)

# Mengambil 29 butir pernyataan Likert
butir = pd.concat([
    df.iloc[:, 6:20],
    df.iloc[:, 21:36]
], axis=1)

print("Jumlah responden :", len(df))
print("Jumlah butir     :", butir.shape[1])
print("Data kosong      :", int(butir.isna().sum().sum()))

display(butir.head())
```

```python
skala = {
    "Sangat Tidak Setuju (STS)": 1,
    "Tidak Setuju (TS)": 2,
    "Netral (N)": 3,
    "Setuju (S)": 4,
    "Sangat Setuju (SS)": 5
}

num = butir.replace(skala).apply(
    pd.to_numeric, errors="coerce"
)

print("Sel gagal dikonversi:",
      int(num.isna().sum().sum()))

display(num.head())
```

```python
dimensi = {}

for kolom in butir.columns:
    nama = re.match(r"^(.*?)\s*\[", kolom.strip()).group(1)
    nama = nama.strip().rstrip(".")
    dimensi.setdefault(nama, []).append(kolom)

for nama, daftar in dimensi.items():
    print(f"{nama}: {len(daftar)} butir")
```

```python
n = len(num)

t = stats.t.ppf(0.975, n - 2)
r_tabel = t / np.sqrt((n - 2) + t**2)

print(f"Jumlah responden = {n}")
print(f"r tabel (α=0,05) = {r_tabel:.3f}")

hasil_validitas = []

for nama, daftar in dimensi.items():
    x = num[daftar]
    total = x.sum(axis=1)

    for i, kolom in enumerate(daftar, 1):
        r, p = stats.pearsonr(x[kolom], total)
        r_koreksi, _ = stats.pearsonr(
            x[kolom], total - x[kolom]
        )

        hasil_validitas.append({
            "Dimensi": nama,
            "Butir": f"B{i}",
            "r hitung": round(r, 3),
            "Corrected r": round(r_koreksi, 3),
            "p-value": round(p, 4),
            "Keterangan": "Valid" if r > r_tabel else "Tidak valid"
        })

tabel_validitas = pd.DataFrame(hasil_validitas)

display(tabel_validitas)
```

```python
def cronbach_alpha(x):
    k = x.shape[1]

    return (k / (k - 1)) * (
        1 - x.var(ddof=1).sum() /
        x.sum(axis=1).var(ddof=1)
    )

def kategori(a):
    return (
        "Sangat baik" if a >= 0.90
        else "Baik" if a >= 0.80
        else "Reliabel" if a >= 0.70
        else "Tidak reliabel"
    )

hasil_reliabilitas = []
alpha_dihapus = []

for nama, daftar in dimensi.items():
    x = num[daftar]

    a = cronbach_alpha(x)

    hasil_reliabilitas.append({
        "Dimensi": nama,
        "Jumlah butir": len(daftar),
        "Cronbach's alpha": round(a, 3),
        "Kategori": kategori(a)
    })

    for i, kolom in enumerate(daftar, 1):
        if len(daftar) > 2:
            alpha_dihapus.append({
                "Dimensi": nama,
                "Butir": f"B{i}",
                "Alpha jika dihapus": round(
                    cronbach_alpha(x.drop(columns=kolom)), 3
                )
            })

tabel_reliabilitas = pd.DataFrame(hasil_reliabilitas)
tabel_alpha_dihapus = pd.DataFrame(alpha_dihapus)

display(tabel_reliabilitas)
display(tabel_alpha_dihapus)
```

```python
gabung = tabel_validitas.merge(
    tabel_alpha_dihapus,
    on=["Dimensi", "Butir"],
    how="left"
)

with pd.ExcelWriter(
    "hasil_uji_validitas_reliabilitas.xlsx"
) as w:
    gabung.to_excel(
        w, sheet_name="Validitas", index=False
    )

    tabel_reliabilitas.to_excel(
        w, sheet_name="Reliabilitas", index=False
    )

files.download("hasil_uji_validitas_reliabilitas.xlsx")
```

```python
# Memilih dimensi Penggunaan AI
nama_dimensi = [
    nama for nama in dimensi
    if nama.startswith("Penggunaan AI")
][0]

daftar = dimensi[nama_dimensi]

# Alpha sebelum penghapusan
x_awal = num[daftar]
alpha_sebelum = cronbach_alpha(x_awal)

# Menghapus butir B4
kolom_hapus = daftar[3]
x_baru = x_awal.drop(columns=kolom_hapus)

# Alpha setelah penghapusan
alpha_sesudah = cronbach_alpha(x_baru)

print("Butir yang dihapus: B4")
print("Alpha sebelum :", round(alpha_sebelum, 3))
print("Alpha sesudah :", round(alpha_sesudah, 3))
print("Perubahan alpha:", round(
    alpha_sesudah - alpha_sebelum, 3
))
```

```python
# Mengambil sampel acak sebanyak 20 responden
num_sample = num.sample(n=20, random_state=1)

# Menghitung r tabel untuk n = 20
n_sample = len(num_sample)
t_sample = stats.t.ppf(0.975, n_sample - 2)

r_tabel_sample = t_sample / np.sqrt(
    (n_sample - 2) + t_sample**2
)

print("Jumlah sampel:", n_sample)
print("r tabel:", round(r_tabel_sample, 3))

# Uji validitas pada sampel baru
hasil_sample = []

for nama, daftar in dimensi.items():
    x = num_sample[daftar]
    total = x.sum(axis=1)

    for i, kolom in enumerate(daftar, 1):
        r, p = stats.pearsonr(x[kolom], total)

        hasil_sample.append({
            "Dimensi": nama,
            "Butir": f"B{i}",
            "r hitung": round(r, 3),
            "Keterangan": (
                "Valid" if r > r_tabel_sample
                else "Tidak valid"
            )
        })

tabel_sample = pd.DataFrame(hasil_sample)

display(tabel_sample)

# Menghitung jumlah butir valid dan tidak valid
print("Jumlah butir valid:",
      (tabel_sample["Keterangan"] == "Valid").sum())

print("Jumlah butir tidak valid:",
      (tabel_sample["Keterangan"] == "Tidak valid").sum())
```

```python
print("LATIHAN 1 - INTERPRETASI VALIDITAS")
print("Nilai r tabel:", round(r_tabel, 3))

tidak_valid = tabel_validitas[
    tabel_validitas["Keterangan"] == "Tidak valid"
]

if len(tidak_valid) == 0:
    print("Tidak ditemukan butir yang tidak valid.")
else:
    display(tidak_valid[
        ["Dimensi", "Butir", "r hitung"]
    ])

print("Jumlah butir valid:",
      (tabel_validitas["Keterangan"] == "Valid").sum())
print("Jumlah butir tidak valid:", len(tidak_valid))

print("\nKesimpulan: Seluruh butir memenuhi kriteria validitas."
      if len(tidak_valid) == 0 else
      "\nKesimpulan: Terdapat butir yang tidak valid.")
```

```python
print("LATIHAN 2 - PERBANDINGAN KRITERIA")

valid_r = (tabel_validitas["r hitung"] > r_tabel).sum()

valid_corrected = (
    tabel_validitas["Corrected r"] >= 0.30
).sum()

print("Valid berdasarkan r hitung:", valid_r)
print("Valid berdasarkan Corrected r:", valid_corrected)
print("Selisih:", abs(valid_r - valid_corrected))
```
