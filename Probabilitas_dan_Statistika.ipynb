```python
import pandas as pd

dataraw = pd.read_excel(
    "rasyid.xlsx",
    header=7,
    usecols="B:H"
)

print("Data berhasil dibaca!")
display(dataraw.head())
```

```python
datafrq = pd.crosstab(
    index=dataraw["Grade"],
    columns="Frekuensi"
)

print("Tabel Frekuensi Grade")
display(datafrq)
```

```python
print("Jumlah data:", len(dataraw))
print("Jumlah kolom:", len(dataraw.columns))

print("\nNama kolom:")
print(dataraw.columns.tolist())
```

```python
plt.figure(figsize=(8,5))

datafrq["Frekuensi"].plot(
    kind="line",
    marker="o"
)

plt.title("Grafik Garis Frekuensi Grade")
plt.xlabel("Grade")
plt.ylabel("Frekuensi")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8,5))

datafrq["Frekuensi"].plot(
    kind="bar"
)

plt.title("Grafik Batang Frekuensi Grade")
plt.xlabel("Grade")
plt.ylabel("Frekuensi")
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(7,7))

datafrq["Frekuensi"].plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Persentase Grade Mahasiswa")
plt.ylabel("")
plt.show()
```

```python
nilai = dataraw["Final Score"].dropna()

print("Jumlah data nilai:", len(nilai))
```

```python
mean = nilai.mean()
median = nilai.median()
modus = nilai.mode().tolist()

print("Mean   :", round(mean, 2))
print("Median :", round(median, 2))
print("Modus  :", modus)
```

```python
minimum = nilai.min()
maksimum = nilai.max()

print("Nilai Minimum :", round(minimum, 2))
print("Nilai Maximum :", round(maksimum, 2))
```

```python
q1 = nilai.quantile(0.25)
q2 = nilai.quantile(0.50)
q3 = nilai.quantile(0.75)

print("Q1 :", round(q1, 2))
print("Q2 :", round(q2, 2))
print("Q3 :", round(q3, 2))
```

```python
variance = nilai.var()
std = nilai.std()

print("Variance :", round(variance, 2))
print("Standard Deviation :", round(std, 2))
```

```python
skewness = nilai.skew()
kurtosis = nilai.kurtosis()
standard_error = std / (len(nilai) ** 0.5)

print("Skewness :", round(skewness, 2))
print("Kurtosis :", round(kurtosis, 2))
print("Standard Error :", round(standard_error, 2))
```
